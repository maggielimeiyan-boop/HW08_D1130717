題目說明：本程式負責處理含有文字格式的原始餐費與人數資料，進行字串清理、型別轉換，並依照「計算服務費 -> 扣除折價券 -> 平均分攤 -> 判斷是否超支」的規則計算出聚餐各項費用與預算狀態。

```python
# ==========================================
# 作業名稱：Python資料型別實作作業：聚餐費用試算器
# 說明：示範字串清理、型別轉換、數值計算、布林判斷與 f-string 格式化輸出
# ==========================================

print("=== 測試一：原始資料 (人數=4, 折扣=100) ===")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 100
budget_per_person = 300

# 1. 整理聚餐名稱 (去除頭尾空白、轉小寫)
clean_event_name = event_name.strip().lower()

# 2. 整理餐費和轉換類型
# 去除 "NT$", ",", " "
clean_meal_str = meal_amount.replace("NT$", "").replace(",", "").strip()
clean_meal_amount = float(clean_meal_str)

# 將人數轉成整數
clean_people = int(people)

# 3. 分步計算費用
service_fee = clean_meal_amount * service_rate
total_amount = clean_meal_amount + service_fee - discount
per_person_cost = total_amount / clean_people

# 4. 判斷是否超過預算 (大於預算為 True，否則為 False)
is_over_budget = per_person_cost > budget_per_person

# 5. 輸出費用摘要 (使用 f-string 格式化到小數點後兩位)
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{clean_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{clean_people} 人")
print(f"每人費用：{per_person_cost:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}\n")
```
